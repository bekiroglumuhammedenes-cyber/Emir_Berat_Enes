## Portföy Optimizasyonu
Portföy optimizasyonu için rastgele portföyler oluşturup, her bir portföyün beklenen getirisini ve riskini hesaplayalım. Daha sonra, bu portföylerin performansını değerlendirelim ve etkin sınırı (efficient frontier) bulalım.

```python
# Ortalama getiri ve kovaryans matrisini hesaplayalım
mean_returns = big_tech_stock_prices.groupby('stock_symbol')['daily_return'].mean()
cov_matrix = big_tech_stock_prices.pivot_table(index='date', columns='stock_symbol', values='daily_return').cov()
```

```python
num_portfolios = 10000
results = np.zeros((4, num_portfolios))
weights_record = []

np.random.seed(42)

for i in range(num_portfolios):
    weights = np.random.random(len(mean_returns))
    weights /= np.sum(weights)
    weights_record.append(weights)
    portfolio_return = np.dot(weights, mean_returns)
    portfolio_stddev = np.sqrt(np.dot(weights.T, np.dot(cov_matrix, weights)))
    results[0, i] = portfolio_return
    results[1, i] = portfolio_stddev
    results[2, i] = results[0, i] / results[1, i]  # Sharpe Ratio
    results[3, i] = i
```

```python
# Sonuçları bir DataFrame'e aktaralım
results_frame = pd.DataFrame(results.T, columns=['Return', 'Risk', 'Sharpe Ratio', 'Index'])
```

```python
# En yüksek Sharpe oranına sahip portföyü bulalım
max_sharpe_idx = results_frame['Sharpe Ratio'].idxmax()
max_sharpe_portfolio = results_frame.loc[max_sharpe_idx]
max_sharpe_weights = weights_record[int(max_sharpe_portfolio['Index'])]
```

```python
# En düşük riske sahip portföyü bulalım
min_risk_idx = results_frame['Risk'].idxmin()
min_risk_portfolio = results_frame.loc[min_risk_idx]
min_risk_weights = weights_record[int(min_risk_portfolio['Index'])]
```

```python
# Etkin sınırı görselleştirelim
plt.figure(figsize=(10, 6))
plt.scatter(results_frame['Risk'], results_frame['Return'], c=results_frame['Sharpe Ratio'], cmap='viridis')
plt.colorbar(label='Sharpe Ratio')
plt.scatter(max_sharpe_portfolio['Risk'], max_sharpe_portfolio['Return'], marker='*', color='r', s=200, label='Max Sharpe Ratio')
plt.scatter(min_risk_portfolio['Risk'], min_risk_portfolio['Return'], marker='*', color='b', s=200, label='Min Risk')
plt.title('Portfolio Optimization based on Efficient Frontier')
plt.xlabel('Risk (Standard Deviation)')
plt.ylabel('Return')
plt.legend()
plt.show()
```

```python
# Maksimum Sharpe oranına sahip portföyün ağırlıklarını görüntüleyelim
print("Maximum Sharpe Ratio Portfolio Allocation\n")
print("Return:", max_sharpe_portfolio['Return'])
print("Risk:", max_sharpe_portfolio['Risk'])
print("Sharpe Ratio:", max_sharpe_portfolio['Sharpe Ratio'])
print("\nWeights:\n")
for i, txt in enumerate(mean_returns.index):
    print(f"{txt}: {max_sharpe_weights[i]}")
```

```python
# Minimum riske sahip portföyün ağırlıklarını görüntüleyelim
print("\nMinimum Risk Portfolio Allocation\n")
print("Return:", min_risk_portfolio['Return'])
print("Risk:", min_risk_portfolio['Risk'])
print("\nWeights:\n")
for i, txt in enumerate(mean_returns.index):
    print(f"{txt}: {min_risk_weights[i]}")
```

## Anomali Tespiti
Zaman serisi verilerinde anormallikleri tespit etmek için çeşitli yöntemler kullanabiliriz. Örneğin, z-score veya IQR yöntemlerini kullanarak anormallikleri belirleyebiliriz. Aşağıda z-score yöntemini kullanarak anomali tespiti yapacağız.

```python
# Anomali tespiti: günlük getiri üzerinden z-score
# DÜZELTME: eski kod fiyat seviyesine z-score uyguluyordu, bu sadece yüksek fiyatlı günleri işaretliyordu.
# Eşik (|z| > 4) bizim seçimimiz.
anomali_ozet = {}
for symbol in unique_symbols:
    stock_data = (big_tech_stock_prices[big_tech_stock_prices['stock_symbol'] == symbol]
                  .sort_values('date').set_index('date'))
    ret = stock_data['close'].pct_change().dropna()
    z = (ret - ret.mean()) / ret.std()
    anom = z.abs() > 4
    anomali_ozet[symbol] = int(anom.sum())

    # Fiyat grafiği üzerinde anomali günlerini işaretleyelim
    plt.figure(figsize=(14, 7))
    plt.plot(stock_data.index, stock_data['close'], label='Close Price')
    plt.scatter(ret[anom].index, stock_data.loc[ret[anom].index, 'close'], color='red', s=15, label='Anomaly (return)')
    plt.title(f'{symbol} Stock Price with Return Anomalies')
    plt.xlabel('Date')
    plt.ylabel('Close Price')
    plt.legend()
    plt.show()

    print(f"{symbol}: {anom.sum()} anomali günü, ilk beşi:", list(ret[anom].index.strftime('%Y-%m-%d')[:5]))

print("\nŞirket başına anomali sayısı:", anomali_ozet)
```

```python
from arch import arch_model

# Her bir hisse senedi için volatilite modellemesi yapalım
for symbol in unique_symbols:
    stock_data = big_tech_stock_prices[big_tech_stock_prices['stock_symbol'] == symbol].copy()
    stock_data.set_index('date', inplace=True)

    # Günlük getiriyi hesaplayalım
    stock_data['return'] = stock_data['close'].pct_change().dropna()

    # GARCH(1,1) modeli
    model = arch_model(stock_data['return'].dropna(), vol='Garch', p=1, q=1)
    model_fit = model.fit(disp='off')
    print(f"Summary for {symbol}:")
    print(model_fit.summary())

    # Volatilite tahminleri
    volatility = model_fit.conditional_volatility

    # Volatiliteyi görselleştirelim
    plt.figure(figsize=(14, 7))
    plt.plot(volatility)
    plt.title(f'{symbol} Stock Volatility')
    plt.xlabel('Date')
    plt.ylabel('Volatility')
    plt.show()

    # Gelecekteki volatiliteyi tahmin edelim
    forecast_horizon = 30
    forecast = model_fit.forecast(horizon=forecast_horizon)
    forecast_volatility = np.sqrt(forecast.variance.values[-1, :])

    plt.figure(figsize=(14, 7))
    plt.plot(range(1, forecast_horizon+1), forecast_volatility)
    plt.title(f'{symbol} Forecasted Volatility for Next 30 Days')
    plt.xlabel('Days')
    plt.ylabel('Volatility')
    plt.show()
```

## Otomatik Alım Satım Algoritmaları (Tüm Şirketler)
Basit bir hareketli ortalama crossover stratejisi kullanarak alım-satım sinyalleri oluşturacağız.

```python
# Her bir hisse senedi için alım-satım sinyalleri oluşturmak
for symbol in unique_symbols:
    stock_data = big_tech_stock_prices[big_tech_stock_prices['stock_symbol'] == symbol].copy()
    stock_data.set_index('date', inplace=True)

    # Hareketli ortalamaları hesaplayalım
    stock_data['SMA50'] = stock_data['close'].rolling(window=50).mean()
    stock_data['SMA200'] = stock_data['close'].rolling(window=200).mean()

    # Alım-satım sinyalleri
    stock_data['Signal'] = 0.0
    stock_data.loc[stock_data.index[50:], 'Signal'] = np.where(
        stock_data['SMA50'].iloc[50:] > stock_data['SMA200'].iloc[50:], 1.0, 0.0)
    stock_data['Position'] = stock_data['Signal'].diff()

    # Sinyalleri görselleştirelim
    plt.figure(figsize=(14, 7))
    plt.plot(stock_data['close'], label='Close Price')
    plt.plot(stock_data['SMA50'], label='50-day SMA', alpha=0.7)
    plt.plot(stock_data['SMA200'], label='200-day SMA', alpha=0.7)
    plt.plot(stock_data[stock_data['Position'] == 1].index, stock_data['SMA50'][stock_data['Position'] == 1], '^', markersize=10, color='g', lw=0, label='Buy Signal')
    plt.plot(stock_data[stock_data['Position'] == -1].index, stock_data['SMA50'][stock_data['Position'] == -1], 'v', markersize=10, color='r', lw=0, label='Sell Signal')
    plt.title(f'{symbol} - SMA Crossover Strategy')
    plt.xlabel('Date')
    plt.ylabel('Close Price')
    plt.legend()
    plt.show()
```

## Bizim eklediğimiz bölüm: hata metrikleri ve zamana göre bölme
Bu hücreler orijinal notebook'ta yoktu, ekip olarak ekledik.

```python
# Hücre B0: metrik fonksiyonları
def mape(y_true, y_pred):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    return np.mean(np.abs((y_true - y_pred) / y_true)) * 100

def metrikler(y_true, y_pred, ad):
    return {
        "Model": ad,
        "RMSE": mean_squared_error(y_true, y_pred) ** 0.5,
        "MAE": mean_absolute_error(y_true, y_pred),
        "MAPE (%)": mape(y_true, y_pred),
        "R²": r2_score(y_true, y_pred),
    }
```

```python
# Hücre B1: tarihe göre bölme (regresyon) + baseline
from sklearn.dummy import DummyRegressor

FEATURES = ['unemployment_rate', 'cpi', 'inflation_rate',
            'mortgage_interest_rate', 'corporate_bond_yield']

df = merged_data.sort_values('date')
KESIM = '2020-01-01'   # kendi seçimimiz; nedenini sunumda söyleyin
train = df[df['date'] < KESIM]
test  = df[df['date'] >= KESIM]

print("Eğitim:", len(train), "satır | Test:", len(test), "satır")
print("Eğitim aralığı:", train['date'].min().date(), "→", train['date'].max().date())
print("Test aralığı  :", test['date'].min().date(), "→", test['date'].max().date())

X_train, y_train = train[FEATURES], train['close']
X_test,  y_test  = test[FEATURES],  test['close']

sonuclar = []
base = DummyRegressor(strategy='mean').fit(X_train, y_train)
sonuclar.append(metrikler(y_test, base.predict(X_test), "Baseline (eğitim ortalaması)"))

lr = LinearRegression().fit(X_train, y_train)
sonuclar.append(metrikler(y_test, lr.predict(X_test), "Doğrusal Regresyon"))

rf = RandomForestRegressor(n_estimators=100, random_state=42).fit(X_train, y_train)
sonuclar.append(metrikler(y_test, rf.predict(X_test), "Random Forest"))

pd.DataFrame(sonuclar).round(3)
```

```python
# Hücre B1b: eğitim / doğrulama / test ayrımı (zamana göre)
from sklearn.dummy import DummyRegressor
from sklearn.base import clone
# Eğitim: 2020 öncesi verinin ilk kısmı, doğrulama: 2018-2019, test: 2020 sonrası (B1 ile aynı test seti).
# Doğrulama seti hiperparametre seçmek için kullanılır; test seti yalnızca en sonda bir kez değerlendirilir.
DOGRULAMA_BASI = '2018-01-01'   # kendi seçimimiz; nedenini raporda yazın
KESIM          = '2020-01-01'   # B1 ile aynı

df = merged_data.sort_values('date')
tr  = df[df['date'] <  DOGRULAMA_BASI]
val = df[(df['date'] >= DOGRULAMA_BASI) & (df['date'] < KESIM)]
te  = df[df['date'] >= KESIM]

for ad, d in [("Eğitim", tr), ("Doğrulama", val), ("Test", te)]:
    print(f"{ad:10s}: {len(d):6d} satır | {d['date'].min().date()} → {d['date'].max().date()}")

# 1) Adaylar: yalnızca eğitim setiyle eğitilir, doğrulama setinde karşılaştırılır
adaylar = {
    "Baseline (eğitim ortalaması)": DummyRegressor(strategy='mean'),
    "Doğrusal Regresyon":           LinearRegression(),
    "RF (varsayılan)":              RandomForestRegressor(n_estimators=100, random_state=42),
    "RF (max_depth=5)":             RandomForestRegressor(n_estimators=100, max_depth=5, random_state=42),
    "RF (min_samples_leaf=200)":    RandomForestRegressor(n_estimators=100, min_samples_leaf=200, random_state=42),
}
dogrulama_sonuc = []
for ad, m in adaylar.items():
    m.fit(tr[FEATURES], tr['close'])
    dogrulama_sonuc.append(metrikler(val['close'], m.predict(val[FEATURES]), ad))
dogrulama_df = pd.DataFrame(dogrulama_sonuc).round(3)
print("\nDoğrulama seti sonuçları:")
display(dogrulama_df)

# 2) Doğrulama RMSE'si en düşük modeli seç (baseline hariç)
en_iyi_ad = dogrulama_df[~dogrulama_df['Model'].str.startswith('Baseline')].sort_values('RMSE').iloc[0]['Model']
print("Doğrulamada seçilen model:", en_iyi_ad)

# 3) Seçilen modeli eğitim+doğrulama ile yeniden eğit, testte bir kez değerlendir
trval = pd.concat([tr, val])
final_sonuc = []
for ad in ["Baseline (eğitim ortalaması)", en_iyi_ad]:
    m = clone(adaylar[ad]).fit(trval[FEATURES], trval['close'])
    final_sonuc.append(metrikler(te['close'], m.predict(te[FEATURES]), ad + " [eğitim+doğrulama]"))
print("\nTest seti sonuçları (yalnızca bir kez):")
display(pd.DataFrame(final_sonuc).round(3))
```

```python
# Hücre B2: rastgele bölme ile tarihe göre bölmeyi yan yana koy
Xr_tr, Xr_te, yr_tr, yr_te = train_test_split(df[FEATURES], df['close'],
                                              test_size=0.2, random_state=42)
rf_r = RandomForestRegressor(n_estimators=100, random_state=42).fit(Xr_tr, yr_tr)
pd.DataFrame([
    metrikler(yr_te, rf_r.predict(Xr_te), "RF - rastgele bölme"),
    metrikler(y_test, rf.predict(X_test), "RF - tarihe göre bölme"),
]).round(3)
```

```python
# Hücre B3: ARIMA test hatası (Apple, son 60 gün test)
TEST_GUN = 60
fiyat = apple_stock['close'].reset_index(drop=True)
arima_train, arima_test = fiyat[:-TEST_GUN], fiyat[-TEST_GUN:]

fit = ARIMA(arima_train, order=(5, 1, 0)).fit()

# (a) 60 gün ileriye tek seferde tahmin
coklu = fit.get_forecast(steps=TEST_GUN).predicted_mean
# (b) Her gün bir sonraki günü tahmin (gerçek veri geldikçe güncellenir)
tek_adim = fit.append(arima_test, refit=False).predict(
    start=len(arima_train), end=len(fiyat) - 1)
# (c) Naive baseline: yarın = bugün
naive = fiyat.shift(1)[-TEST_GUN:]

pd.DataFrame([
    metrikler(arima_test, coklu,    "ARIMA (60 gün ileri)"),
    metrikler(arima_test, tek_adim, "ARIMA (1 gün ileri)"),
    metrikler(arima_test, naive,    "Naive (dünün fiyatı)"),
]).round(3)
```

```python
# Hücre B4 (düzeltilmiş): tüm şirketler için ARIMA test hatası, adil karşılaştırma
# 60 günlük tahmin ARIMA ile naive'in 60 günlük hali (son eğitim fiyatını sabit tutmak) karşılaştırılır.
# 1 günlük tahmin de ARIMA ile naive (dünün fiyatı) arasında ayrıca karşılaştırılır.
satirlar = []
for sembol in big_tech_stock_prices['stock_symbol'].unique():
    s = (big_tech_stock_prices[big_tech_stock_prices['stock_symbol'] == sembol]
         .sort_values('date')['close'].reset_index(drop=True))
    tr, te = s[:-TEST_GUN], s[-TEST_GUN:]
    f = ARIMA(tr, order=(5, 1, 0)).fit()
    arima60 = f.get_forecast(steps=TEST_GUN).predicted_mean
    arima1  = f.append(te, refit=False).predict(start=len(tr), end=len(s) - 1)
    naive60 = np.repeat(tr.iloc[-1], TEST_GUN)
    naive1  = s.shift(1)[-TEST_GUN:]
    satirlar.append({
        "Sembol": sembol,
        "ARIMA 60g MAPE (%)": mape(te, arima60),
        "Naive 60g MAPE (%)": mape(te, naive60),
        "ARIMA 1g MAPE (%)": mape(te, arima1),
        "Naive 1g MAPE (%)": mape(te, naive1),
    })
pd.DataFrame(satirlar).round(2)
```
